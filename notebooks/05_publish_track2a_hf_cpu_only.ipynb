## 1. Clone latest verified code and check package requirements

```python
import pathlib, subprocess, sys, os
ROOT=pathlib.Path('/content/PolicyProof-AI')
if not ROOT.exists():
    subprocess.run(['git','clone','--depth','1','https://github.com/betterlifestyle04-commits/PolicyProof-AI.git',str(ROOT)],check=True)
else:
    subprocess.run(['git','-C',str(ROOT),'pull','--ff-only'],check=True)
sys.path.insert(0,str(ROOT))
subprocess.run([sys.executable,'-m','pip','-q','install','huggingface_hub>=0.36,<2'],check=True)
print('Source project prepared. No network publication attempted.')
```

## 2. Upload **one verified local ZIP**, then run an offline integrity check

Use Colab's picker to upload the research ZIP. Do **not** upload your model token, personal files or the original gated model.

```python
from google.colab import files
import pathlib
uploaded=files.upload()
assert len(uploaded)==1, 'Upload exactly one verified Track 2a dataset ZIP.'
ARCHIVE=pathlib.Path('/content')/next(iter(uploaded))
assert ARCHIVE.name=='PolicyProof_Track2a_HF_Dataset_Upload_Ready_NOT_PUBLISHED.zip', 'Wrong archive name'
from scripts.publish_track2a_dataset import validate_archive
validated=validate_archive(ARCHIVE)
print('Offline integrity validated:',len(validated),'files; 48 scored, 72 UNSCORED. No HF actions yet.')
```

## 3. Choose a dataset license and perform a **dry run**

Set `DATASET_LICENSE` only after independently reviewing your publishing rights. Allowed identifiers: `cc-by-4.0`, `cc0-1.0`, `cc-by-sa-4.0`. An empty value will stop rather than guess a license.

```python
DATASET_LICENSE = ''  # OWNER MUST CHOOSE AND APPROVE. Leave blank to stop.
assert DATASET_LICENSE in {'cc-by-4.0','cc0-1.0','cc-by-sa-4.0'}, 'Choose an approved license after checking rights'
subprocess.run([sys.executable,'-m','scripts.publish_track2a_dataset',
    '--archive',str(ARCHIVE),'--license',DATASET_LICENSE],cwd=ROOT,check=True)
print('Offline publication preflight OK. No external upload.')
```

## 4. OPTIONAL: publish a public dataset and collection **only with explicit approval**

Create a **write-scoped Hugging Face token** and save it in Colab Secrets as `HF_TOKEN`; grant this notebook access. **Do not paste the token into the notebook or chat.** Set `HF_DATASET_REPO_ID` to your actual `username/dataset-slug`. Change `PUBLISH_DATASET` to `True` and `I_CONFIRM_RIGHTS_AND_PUBLICATION` to `True` only when you knowingly choose to publish the synthetic data publicly. This step creates a public HF dataset repository and a public dataset collection, then verifies anonymous visibility. It does *not* create the separate model collection requested by Lexsi.

```python
PUBLISH_DATASET=False
I_CONFIRM_RIGHTS_AND_PUBLICATION=False
HF_DATASET_REPO_ID=''  # Example format username/policyproof-track2a-safety-2026
if not PUBLISH_DATASET:
    print('Publication disabled — no dataset or collection created.')
else:
    assert I_CONFIRM_RIGHTS_AND_PUBLICATION is True, 'Owner must approve rights and public publication'
    assert DATASET_LICENSE in {'cc-by-4.0','cc0-1.0','cc-by-sa-4.0'}, 'License approval missing'
    assert '/' in HF_DATASET_REPO_ID, 'Enter your actual Hugging Face namespace/dataset'
    from google.colab import userdata
    token=userdata.get('HF_TOKEN')
    assert token, 'Set HF_TOKEN securely under Colab Secrets; do not paste it into cells'
    env=os.environ.copy()
    env['HF_TOKEN']=token
    subprocess.run([sys.executable,'-m','scripts.publish_track2a_dataset',
      '--archive',str(ARCHIVE),'--license',DATASET_LICENSE,
      '--repo-id',HF_DATASET_REPO_ID,'--publish','--confirm-rights',
      '--confirm-publication','--create-collection',
      '--receipt','/content/track2a_hf_publication_receipt.json'],
      cwd=ROOT,env=env,check=True)
    print('Dataset publication check complete. Read the JSON receipt; it does not mean Lexsi submission is finished.')
```
